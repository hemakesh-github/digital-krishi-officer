```python
import pandas as pd
import re
import os

# Load dataset
file_path = r'c:\Documents\farmerAssist\kcc_dataset.csv'
try:
    df = pd.read_csv(file_path)
except Exception as e:
    print(f"Error reading CSV: {e}")
    # Try with different encoding if needed
    try:
        df = pd.read_csv(file_path, encoding='latin1')
    except Exception as e2:
        print(f"Error reading CSV with latin1: {e2}")
        exit(1)
```

```python
df_clean = df.drop_duplicates()

# Preprocessing / Normalization
def normalize_crop(crop_name):
    if not isinstance(crop_name, str):
        return "Unknown"
    crop_name = crop_name.lower().strip()
    if 'paddy' in crop_name or 'rice' in crop_name:
        return 'Rice'
    if 'wheat' in crop_name:
        return 'Wheat'
    if 'cotton' in crop_name:
        return 'Cotton'
    if 'maize' in crop_name:
        return 'Maize'
    if 'chilli' in crop_name or 'chillies' in crop_name:
        return 'Chilli'
    if 'sugarcane' in crop_name:
        return 'Sugarcane'
    if 'groundnut' in crop_name:
        return 'Groundnut'
    print(crop_name)    
    return crop_name.title()

df_clean['Crop_Normalized'] = df_clean['Crop'].apply(normalize_crop)
```

```python
# Normalize Query Type
def normalize_query_type(q_type):
    if not isinstance(q_type, str):
        return "Other"
    q_type = q_type.lower().strip()
    if 'weather' in q_type:
        return 'Weather'
    if 'market' in q_type:
        return 'Market Info'
    if 'scheme' in q_type:
        return 'Government Schemes'
    if 'plant protection' in q_type or 'pest' in q_type or 'disease' in q_type:
        return 'Plant Protection'
    if 'nutrient' in q_type or 'fertilizer' in q_type:
        return 'Nutrient Management'
    return q_type.title()

df_clean['QueryType_Normalized'] = df_clean['QueryType'].apply(normalize_query_type)
```
